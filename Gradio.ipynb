```python
import gradio as gr
def greet(name):
    return "Hello " + name + "!"
demo = gr.Interface(fn=greet, inputs="text", outputs="text")
demo.launch()   
```

```python
import gradio as gr
from PIL import Image

def generate_image(prompt):
    img=Image.open(r"C:\Users\Neha\OneDrive\Desktop\Img1.jpg")
    return img
    
with gr.Blocks() as demo:
    gr.Markdown("# Fancy Image Generator")
    with gr.Row(equal_height=True):
        textbox = gr.Textbox(lines=1,show_label=False,placeholder="Enter Prompt...")
        button = gr.Button("Generate",variant="primary")
    image= gr.Image(type="pil")

    button.click(
        fn=generate_image,
        inputs=textbox,
        outputs=image,
    )
demo.launch()
```
