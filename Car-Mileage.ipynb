```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
%matplotlib inline
from sklearn import preprocessing
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.metrics import r2_score
```

```python
dataset=pd.read_csv(r'C:\Users\Neha\Downloads\AI\18th- l1, l2, scaling\lasso, ridge, elastic net\TASK-22_LASSO,RIDGE\car-mpg.csv')
```

```python
dataset.head(15)
```

```python
dataset['origin'] = dataset['origin'].replace({1: 'america',2: 'europe',3: 'asia'})
dataset = pd.get_dummies(dataset,columns = ['origin'],dtype=int)
dataset = dataset.replace('?', np.nan)
dataset
```

```python
dataset=dataset.apply(pd.to_numeric, errors='ignore')
numeric_cols=dataset.select_dtypes(include=[np.number]).columns
dataset[numeric_cols]=dataset[numeric_cols].apply(lambda x: x.fillna(x.median()))
```

```python
dataset.head()
```

# Model Building

```python
x = dataset.drop(['mpg'], axis = 1)
y = dataset[['mpg']]
```

```python
x_s = preprocessing.scale(x)
x_s = pd.DataFrame(x_s,columns = x.columns)

y_s = preprocessing.scale(y)
y_s = pd.DataFrame(y_s,columns = y.columns)
```

```python
x_s
```

```python
y_s
```

```python
dataset.shape
```

```python
x_train,x_test,y_train,y_test = train_test_split(x_s,y_s,test_size=0.2,random_state=0)
x_train.shape
```

```python
x_test.shape
```

```python
y_train.shape
```

```python
y_test.shape
```

# Simple Linear Model

```python
regression_model = LinearRegression()
regression_model.fit(x_train,y_train)

for idx, col_name in enumerate(x_train.columns):
    print('The coefficient for {} is {}'.format(col_name, regression_model.coef_[0][idx]))
intercept = regression_model.intercept_[0]
print('The intercept is {}'.format(intercept))
```

# Regularized Ridge Regression

```python
ridge_model = Ridge(alpha=0.4)
ridge_model.fit(x_train,y_train)

print('Ridge model coef: {}'.format(ridge_model.coef_))
```

# Regularized Lasso Regression

```python
lasso_model = Lasso(alpha = 0.1)
lasso_model.fit(x_train,y_train)

print('Lasso model coef: {}'.format(lasso_model.coef_))
```

# Score Comparison

```python
#simple linear model
print(regression_model.score(x_train,y_train))
print(regression_model.score(x_test,y_test))

print('************************************')
#Ridge
print(ridge_model.score(x_train,y_train))
print(ridge_model.score(x_test,y_test))

print('************************************')
#Lasso
print(lasso_model.score(x_train,y_train))
print(lasso_model.score(x_test,y_test))
```
