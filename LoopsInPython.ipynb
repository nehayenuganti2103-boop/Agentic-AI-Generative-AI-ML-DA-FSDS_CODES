# WHILE LOOP

```python
i=1
while i<=5:
    print('Work In Silence.Let Your Success Make The Sound')
    i=i+1  #increment
```

```python
i=5
while i>=1:
    print('Work In Silence.Let Your Success Make The Sound')
    i=i-1  #decrement
```

```python
i=1
while i<=5:
    print('Work In Silence.Let Your Success Make The Sound :',i)  #with position value
    i=i+1  #increment
```

```python
i=5
while i>=1:
    print('Work In Silence.Let Your Success Make The Sound :',i)  #with postion value
    i=i-1  #decrement
```

```python
i=1
while i<=5:
    print('AI Developer')
    j=1
    while j<=4:
        print('Data Scientist')
        j=j+1

    i=i+1
    print()
```

```python
i=1
while i<=5:
    print('AI Developer',end="     ")
    j=1
    while j<=4:
        print('Data Scientist',end="  ")
        j=j+1

    i=i+1
    print()
```

```python
i=1

while i<=2:
    j=0
    while j<=2:
        print(i*j,end=" ")
        j+=1
    print()
    i+=1
```

```python
i=1
while i<=4:
    j=0
    while j<=3:
     print(i*j,end=" ")
     j+=1
    print()
    i+=1
```

```python
i=10

while i<=4:
    j=1
    while j<=3:
        print(i*j,end=" ")
        j+=1

    print()
    i+=1

else:
    print('Number Not Found')
```

# BREAK , CONTINUE AND PASS Conditions 

```python
for i in range(1,11):
    print(i)
```

```python
for i in range(1,11):
    if i==7:
        break         #break the flow
    print(i)
```

```python
for i in range(1,11):
    if i==9:
        continue     #skips the particular value
    print(i)
```

```python
for i in range(1,11):
```

```python
for i in range(1,11):
    pass           #avoids the error as output 
```
