```python
a=10
def Num():
    global a
    b=15
    print('In-Function',b)
    print('Global variable',a)
Num()
print('Out-Function',a)
```

```python
x=10
def update_x():
    global x
    x+=5
update_x()
print(x)
```

```python
x=10
def update_x():
    globals()['x']+=5
update_x()
print(x)
```

```python
def count(list):
    
    even=0
    odd=0

    for i in list:
        if i%2 == 0:
            even += 1
        else:
            odd += 1
    return even,odd
    
list = [1,2,3,4,5,6,7,8,9]
even, odd = count(list)

print(even)
print(odd)
```
