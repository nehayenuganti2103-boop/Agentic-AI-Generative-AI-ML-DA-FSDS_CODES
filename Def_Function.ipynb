# Keyword Argument

```python
def person(name,age):
    print(name)
    print(age+1)
person(name='Neha',age=23)
```

```python
def person(name,age,city):
    print(name)
    print(age)
    print(city)
person(name='Neha',age=24,city='hyd')
```

```python
def person(name,age,city):
    print(name)
    print(age)
    print(city='hyd')
person(name='Neha',age=24)
```

# Random Number Generator

```python
import random
def get_random_number():
    num=random.randint(1,10)
    return num
get_random_number()
```

# Current Date

```python
from datetime import date
def show_current_date():
    d=date.today()
    return d.isoformat()
show_current_date()
```

# Square Of A Number

```python
def calculate_square(num):
    num=num**2
    return num
calculate_square(11)
```

# Area And Perimeter Of A Square

```python
def square_area_perimeter(side):
    a=side*side
    p=4*side
    return f"Area and Perimeter of a square:{a},{p}"
square_area_perimeter(10)
```

# Even Or Odd

```python
def check_even_odd(num):
    if num%2==0:
        return 'Even'
    else:
        return 'Odd'
check_even_odd(19)
```

# ASCII Code Generation

```python
def draw_star():
    for i in range(1,6):
        for j in range(1,i):
            print('*',end=' ')
        print()
draw_star()
```

# Calci Menu

```python
def display_menu():
    l=['Add','Sub','Mul','Div','E']
    for i in l:
        print(i)
display_menu()
```
