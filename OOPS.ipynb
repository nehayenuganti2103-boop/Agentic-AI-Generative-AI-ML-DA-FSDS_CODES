```python
class student(): 
        
    school = 'nit'
    
    def __init__(self, m1, m2, m3) : # hear i pass 3 variable 
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3
#lets create a method to calculate average of the marks 
    
    def avg(self):   # ------> hear this average is INSTANCE METHOD 
        return (self.m1+self.m2+self.m3)/3
    
    @classmethod    # decorator
    def info(cls): # user need to define cls - for class variable 
        return cls.school 
        
# lets create  object hear as we created s1, s2 we created 2 object & also i want to pass value in this  
s1 = student(10, 20, 30)
s2 = student(5, 15, 10)

print(s1.avg())
print(student.info())
```

```python
class student(): 
        
    school = 'nit'
    
    def __init__(self, m1, m2, m3) : # hear i pass 3 variable 
        self.m1 = m1
        self.m2 = m2
        self.m3 = m3
#lets create a method to calculate average of the marks 
    def avg(self):   # ------> hear this average is INSTANCE METHOD 
        return (self.m1+self.m2+self.m3)/3
    
    @classmethod
    def getSchool(cls):
        return cls.school
    
    @staticmethod # this student class nothing to do any class & for that also u need to pass speciasl decorator
    def info():
        print('this is student class') 
        
# lets create  object hear as we created s1, s2 we created 2 object & also i want to pass value in this  
s1 = student(10, 20, 30)
s2 = student(5, 15, 10)

print(s1.avg()) 
print(s2.avg())

print(student.getSchool())

student.info()
```

```python
class A():                      A={feature1,feature2}
    def feature1(self):        
        print('feature1')
    def feature2(self):
        print('feature2')
        
class B():                     B={feature3,feature4}
    def feature3(self):
        print('feature3')
    def feature4(self):
        print('feature4 ')
a = A()
b = B() 
```

```python
## Single-Level-Inheritence
class A():                      #A={feature1,feature2}
    def feature1(self):        
        print('feature1')
    def feature2(self):
        print('feature2')
        
class B(A):                     #B={feature1,feature2,feature3,feature4}
    def feature3(self):
        print('feature3')
    def feature4(self):
        print('feature4 ')
a = A()
b = B() 
```

```python
## Multi-Level Inheritence
class A():                                    #A={feature1,feature2}
    def feature1(self):        
        print('feature1')
    def feature2(self):
        print('feature2')
 
class B():                                   #B={feature3,feature4}
    def feature3(self):
        print('feature3')
    def feature4(self):
        print('feature4')
class C(B):                                 #C={feature3,feature4,feature5}
    def feature5(self):
        print('feature5')
        
a = A()
b = B() 
c = C()
```

```python
## Multiple Inheritence
class A():                             #A={feature1,feature2}
    def feature1(self):         
        print('feature1')
    def feature2(self):
        print('feature2')
 
class B():                            #B={feature3,feature4}
    def feature3(self):
        print('feature3')
    def feature4(self):
        print('feature4')
        
class C(A, B):                       #C={feature1,feature2,feature3,feature4,feature5}
    def feature5(self):
        print('feature5')
        
a = A()
b = B() 
c = C()
```

```python
## Super Function
class A() : 
    def __init__(self):
        print('in the A init')
    def feature1(self):         
        print('feature1')
    def feature2(self):
        print('feature2 is hi')

class B(A):
    def __init__(self):
        super().__init__()  
        print('in the B init')
    def feature3(self):        
        print('feature3')
    def feature4(self):
        print('feature4')
        
a1 = B()
```
