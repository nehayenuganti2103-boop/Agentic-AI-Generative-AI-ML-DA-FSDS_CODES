```python
import matplotlib.pyplot as plt 
from matplotlib import style 

males=(titanic['Sex']==1).sum()
females=(titanic['Sex']==2).sum()
print(males)
print(females)
p=[males,females]
plt.pie(p,
        labels=['Male','Female'],
        colors=['blue','red'],
        explode=(0.15,0),
        startangle=0)
plt.axis('equal')
plt.show()
```

```python
MaleS=titanic[titanic.Sex==1][titanic.Survived==1].shape[0]
print(MaleS)
MaleN=titanic[titanic.Sex==1][titanic.Survived==0].shape[0]
print(MaleN)
FemaleS=titanic[titanic.Sex==2][titanic.Survived==1].shape[0]
print(FemaleS)
FemaleN=titanic[titanic.Sex==2][titanic.Survived==0].shape[0]
print(FemaleN)
```

```python
charts=[MaleS,MaleN,FemaleS,FemaleN]
colors=['lightblue','red','yellow','green']
labels=['Survived Male','Un-Survived Male','Survived Female','Un-Survived Female']
explode=[0,0.05,0,0.1]
plt.pie(charts,labels=labels,colors=colors,explode=explode,startangle=100,counterclock=False,autopct="%.2f%%")
plt.axis("equal")
plt.show()
```

```python
import ollama 
def generate_insights(titanic_summary):
    prompt = f"Analyze the dataset summary and provide insights:\n\n{titanic_summary}"
    response = ollama.chat(model="gemma3:270m",messages=[{"role":"user","content":prompt}])
    return response['message']['content']
#Generate AI insights
summary = titanic.describe().to_string()
insights = generate_insights(summary)
print("\n AI_Generated Insights:\n",insights)
```

```python
import gradio as gr
def eda_analysis(file):
    titanic = pd.read_csv(file.name)
    summary = titanic.describe().to_string()
    insights = generate_insights(summary)
    return insights
#create web interface
demo=gr.Interface(fn=eda_analysis, inputs="file",outputs="text",title="AI-Powered EDA with Gemma3:270m")
#launch APP
demo.launch(share=True)
```
